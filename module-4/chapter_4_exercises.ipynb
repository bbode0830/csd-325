```python
# 4-4

# import the load_workbook "chapter" of the openpyxl "cookbook"
from openpyxl import load_workbook
import csv

# instead of reading the file, we use the load_workbook 'recepie' and store it in a variable
source_workbook = load_workbook(filename = 'U6RATE.xlsx')

# see the different sheets in the workbook
print(source_workbook.sheetnames)

# for each sheet, we will write a csv file
for sheet_num, sheet_name in enumerate(source_workbook.sheetnames):

    current_sheet = source_workbook[sheet_name]
    
    # see what the name of the current sheet is
    print(sheet_name)
    
    # create and open a file for the data in 'write' mode
    output_file = open("xlsx_"+sheet_name+".csv","w")
    
    #use the writer method from the csv library to write the data to a csv file
    output_writer = csv.writer(output_file)

    #go through every row in the sheet
    for row in current_sheet.iter_rows():

        #make an empty list for the values in each cell
        row_cells = []

        # loop through each cell in the row
        for cell in row:

            # print what is in the cell for our own edification
            print(cell, cell.value)

            # add the value of each cell to our empty list of cell values
            row_cells.append(cell.value)

        # write the new row values to the output file
        output_writer.writerow(row_cells)
        
    #close the csv file
    output_file.close()
```

```python
# 4-6 reading data from a .xls file using the xrld library

import xlrd
import csv

# make a variable that stores the file using the open_workbook 'recepie'
source_workbook = xlrd.open_workbook("U6RATE_xls.xls")

# go through each sheet in the workbook
for sheet_name in source_workbook.sheet_names():

    # make a variable for the current sheet using the sheet_by_name recepie
    current_sheet = source_workbook.sheet_by_name(sheet_name)
    print(sheet_name)

    # make an output file
    output_file = open("xls_"+sheet_name+".csv","w")

    # use the csv writer to write into the new csv file
    output_writer = csv.writer(output_file)

    # iterate through each row in the sheet
    for row_num, row in enumerate(current_sheet.get_rows()):

        # use the writerow and row_values recepies to write the values of each row to the new file
        output_writer.writerow(current_sheet.row_values(row_num))

    # close the new file
    output_file.close()
```

```python
# 4-7

import csv

filename = "ghcnd-stations"
source_file = open(filename+".txt","r")

# use the .readlines recepie to read each line and convert to a list of lines
stations_list = source_file.readlines()

output_file = open(filename+".csv","w")
output_writer = csv.writer(output_file)

# make a list of headers based on the readme file
headers = ["ID","LATITUDE","LONGITUDE","ELEVATION","STATE","NAME","GSN_FLAG","HCNCRN_FLAG","WMO_ID"]

# write the headers to the output file
output_writer.writerow(headers)

# loop through each line of the file
for line in stations_list:

    # make an empty list for the current row
    new_row = []

    # add the ID as the fist entry in the list, positions 1-11 in the original file
    new_row.append(line[0:11])
    # latitude
    new_row.append(line[12:20])
    # longitude
    new_row.append(line[21:30])
    # elevation
    new_row.append(line[31:37])
    # state
    new_row.append(line[38:40])
    # name
    new_row.append(line[41:71])
    # gsn flag
    new_row.append(line[72:75])
    # hcncrn flag
    new_row.append(line[76:79])
    # wmo ID
    new_row.append(line[80:85])

    # write the new_row list to the new file
    output_writer.writerow(new_row)

#close the new file
output_file.close()
```
